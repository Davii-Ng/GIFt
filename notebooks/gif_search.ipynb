## 2. Retriever model: text -> 384-dim vector

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
vec = model.encode('a dog being confused')
print(vec.shape)
```

## 3. Similar sentences get similar vectors

```python
from sentence_transformers import util
a = model.encode('the stock market took a turn for the worse')
b = model.encode('the stock market performed worse than expected')
c = model.encode('a cat plays with a ball')
print('similar :', util.cos_sim(a, b).item())
print('different:', util.cos_sim(a, c).item())
```

## 5. Search

```python
from search import search_gif
from IPython.display import HTML, display

def show(query, k=9):
    res = search_gif(query, k)
    figs = ''.join(
        f'<figure style="margin:4px"><img src="{r["url"]}" style="width:120px;height:90px"><figcaption style="width:120px;font-size:10px">{r["description"]}</figcaption></figure>'
        for r in res)
    display(HTML(f'<div style="display:flex;flex-wrap:wrap">{figs}</div>'))

show('a dog being confused')
```
